# Machine Learning với dữ liệu Netflix

Notebook này xây dựng một mô hình phân loại nội dung Netflix thành **Movie** hoặc **TV Show** từ file `netflix_titles_clean.csv`. Đây là **Machine Learning cổ điển có giám sát (supervised learning)**, không phải chatbot hay mô hình sinh nội dung.

Quy trình: **đọc dữ liệu → chọn feature/target → chia train/test → tiền xử lý → train → đánh giá → dự đoán dữ liệu mới**.

> Để chạy lại: chọn kernel Python trong thư mục `.venv`, sau đó chọn **Run All**.

## 1. Bài toán và lựa chọn model

- **Target (`y`)** là đáp án mà model cần học: cột `type`, với `Movie = 0` và `TV Show = 1`.
- **Features (`X`)** là thông tin model được phép nhìn thấy: `description`, `rating`, `country`, `release_year`.
- **Model**: Logistic Regression.

Logistic Regression phù hợp vì đây là bài toán phân loại nhị phân và feature `description` sau TF-IDF có số chiều lớn. Model huấn luyện nhanh, trả về được xác suất và là baseline phù hợp với nội dung Section 7–8.

Không dùng `show_id`, `title`, `duration`, `duration_value`, `duration_unit` hoặc `listed_in`. Đặc biệt, `duration_unit` và những từ như *TV Shows*/*Movies* trong `listed_in` có thể tiết lộ trực tiếp target, gây **data leakage** — model có điểm cao vì nhìn thấy đáp án gián tiếp chứ không thật sự học được quy luật có ích.

### Giải thích các thư viện

- `Path`: tạo và kiểm tra đường dẫn file theo cách an toàn.
- `numpy`: cung cấp `np.nan` cho giá trị thiếu và các phép tính mảng.
- `pandas`: đọc CSV, xử lý bảng dữ liệu và tạo DataFrame.
- `matplotlib.pyplot`: vẽ confusion matrix.
- `display`: hiển thị DataFrame đẹp trong notebook.
- `ColumnTransformer`: áp dụng cách xử lý khác nhau cho từng nhóm cột.
- `TfidfVectorizer`: chuyển văn bản thành vector số.
- `SimpleImputer`: điền giá trị bị thiếu.
- `LogisticRegression`: model phân loại chính.
- `accuracy_score`, `classification_report`, `ConfusionMatrixDisplay`: đánh giá kết quả.
- `train_test_split`, `cross_val_score`: chia dữ liệu và cross-validation.
- `Pipeline`: nối tiền xử lý và model thành một quy trình thống nhất.
- `OneHotEncoder`, `StandardScaler`: biến category thành số và chuẩn hóa số.

`pd.set_option('display.max_columns', None)` chỉ thay đổi cách hiển thị: pandas sẽ không ẩn bớt cột khi in DataFrame.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

pd.set_option('display.max_columns', None)
```

## 2. Đọc và kiểm tra dữ liệu

- `Path(...)` trỏ tới file CSV trong cùng thư mục với notebook.
- Khối `if` dừng chương trình với thông báo rõ ràng nếu thiếu file.
- `pd.read_csv()` đọc toàn bộ CSV thành DataFrame `df`.
- `df.head()` hiển thị 5 dòng đầu để kiểm tra dữ liệu đã đọc đúng.
- `value_counts()` đếm số Movie và TV Show, giúp phát hiện target bị mất cân bằng.

```python
DATA_PATH = Path('netflix_titles_clean.csv')
if not DATA_PATH.exists():
    raise FileNotFoundError(f'Không tìm thấy {DATA_PATH.resolve()}')

df = pd.read_csv(DATA_PATH)
print(f'Số dòng: {len(df):,} | Số cột: {df.shape[1]}')
display(df.head())
display(df['type'].value_counts().rename_axis('Loại').to_frame('Số lượng'))
```

## 3. Chuẩn bị feature và target

Dữ liệu được chia thành 80% train và 20% test. Train set dùng để học; test set được giữ riêng để kiểm tra model trên dữ liệu chưa từng nhìn thấy.

Giải thích code:

1. `features` khai báo đúng bốn cột đầu vào.
2. `.copy()` tạo bảng riêng, tránh vô tình sửa `df` gốc.
3. `fillna('')` thay description thiếu bằng chuỗi rỗng để TF-IDF xử lý được.
4. Chuỗi rỗng trong `rating` và `country` được đổi thành `NaN`; pipeline sẽ điền chúng sau.
5. `pd.to_numeric(..., errors='coerce')` ép `release_year` thành số; giá trị sai định dạng trở thành `NaN`.
6. `X` chứa feature; `y` chứa đáp án và được map thành 0/1 vì model làm việc với số.
7. `test_size=0.20` dành 20% dữ liệu để test.
8. `random_state=42` giúp mỗi lần chạy tạo đúng cùng một phép chia.
9. `stratify=y` giữ tỷ lệ Movie/TV Show gần như giống nhau ở train và test.

```python
features = ['description', 'rating', 'country', 'release_year']
ml_df = df[features + ['type']].copy()

ml_df['description'] = ml_df['description'].fillna('')
ml_df['rating'] = ml_df['rating'].replace('', np.nan)
ml_df['country'] = ml_df['country'].replace('', np.nan)
ml_df['release_year'] = pd.to_numeric(ml_df['release_year'], errors='coerce')

X = ml_df[features]
y = ml_df['type'].map({'Movie': 0, 'TV Show': 1})

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f'Train: {len(X_train):,} dòng | Test: {len(X_test):,} dòng')
display(y_train.map({0: 'Movie', 1: 'TV Show'}).value_counts(normalize=True).rename('Tỷ lệ'))
```

## 4. Tạo pipeline và train model

### 4.1. Ba nhánh tiền xử lý

**Nhánh văn bản:** `TfidfVectorizer` tách `description` thành từ/cụm từ rồi biến chúng thành các cột số. `max_features=5000` giữ tối đa 5.000 đặc trưng; `ngram_range=(1, 2)` dùng cả từ đơn và cụm hai từ; `min_df=2` bỏ từ chỉ xuất hiện một lần; `stop_words='english'` bỏ các từ tiếng Anh quá phổ biến như *the*, *is*, *and*.

**Nhánh categorical:** `SimpleImputer(strategy='most_frequent')` điền giá trị thiếu bằng giá trị phổ biến nhất. `OneHotEncoder` biến mỗi category thành cột 0/1; `handle_unknown='ignore'` giúp dự đoán được category mới chưa xuất hiện trong train set.

**Nhánh số:** giá trị `release_year` thiếu được điền bằng trung vị, sau đó `StandardScaler` đưa cột về thang đo có trung bình gần 0 và độ lệch chuẩn gần 1.

`ColumnTransformer` chạy ba nhánh trên đúng cột của chúng rồi ghép kết quả thành một ma trận feature. `Pipeline` đảm bảo tiền xử lý và model luôn chạy theo đúng thứ tự, đồng thời ngăn test set ảnh hưởng tới quá trình học tiền xử lý.

### 4.2. Tham số của Logistic Regression

- `class_weight='balanced'`: tăng trọng số lỗi của lớp TV Show vì lớp này ít hơn Movie.
- `C=0.7`: `C` càng nhỏ thì regularization càng mạnh. Giá trị 0.7 giảm overfitting nhẹ so với mặc định 1.0.
- `max_iter=2000`: cho thuật toán tối đa 2.000 vòng để hội tụ.
- `random_state=42`: giữ tính tái lập khi thuật toán cần yếu tố ngẫu nhiên.
- `model.fit(X_train, y_train)`: fit toàn bộ bước tiền xử lý trên train set, sau đó học trọng số Logistic Regression.

```python
text_transformer = TfidfVectorizer(
    max_features=5000, ngram_range=(1, 2), min_df=2, stop_words='english'
)

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer(transformers=[
    ('description_tfidf', text_transformer, 'description'),
    ('categorical', categorical_transformer, ['rating', 'country']),
    ('numeric', numeric_transformer, ['release_year'])
])

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        C=0.7, max_iter=2000, class_weight='balanced', random_state=42
    ))
])

model.fit(X_train, y_train)
print('Đã train xong Logistic Regression.')
```

### 4.3. Nguyên lý model hoạt động

#### Bước 1 — Biến dữ liệu thành số

Máy tính không hiểu trực tiếp câu mô tả hay tên quốc gia. Pipeline biến một nội dung thành vector số, ví dụ khái niệm:

`[tfidf_detective, tfidf_episode, rating_TV-MA, country_US, release_year_scaled, ...]`

Với TF-IDF, một từ có trọng số cao khi nó xuất hiện nhiều trong mô tả hiện tại nhưng không xuất hiện quá phổ biến trong toàn bộ dataset. Có thể hiểu gần đúng: **TF-IDF = độ thường xuyên trong văn bản × độ hiếm trong toàn bộ tập dữ liệu**.

#### Bước 2 — Tính điểm tuyến tính

Logistic Regression học một trọng số cho mỗi feature và tính:

$$z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b$$

Feature có trọng số dương đẩy dự đoán về TV Show; trọng số âm đẩy về Movie. Model không dùng một quy tắc viết tay mà tự tìm các trọng số giảm sai số trên train set.

#### Bước 3 — Đổi điểm thành xác suất

Hàm sigmoid nén mọi giá trị $z$ về khoảng 0–1:

$$P(\text{TV Show}) = \frac{1}{1 + e^{-z}}$$

Nếu xác suất từ 0.5 trở lên, `predict()` trả về lớp 1 (TV Show); ngược lại trả về lớp 0 (Movie). `predict_proba()` trả về xác suất của cả hai lớp.

#### Bước 4 — Regularization

Trong lúc học, model vừa cố dự đoán đúng vừa bị phạt nếu trọng số quá lớn. Cơ chế này gọi là regularization và giúp model bớt ghi nhớ chi tiết riêng của train set. `C=0.7` điều khiển mức phạt đó.

## 5. Đánh giá model

Code gọi `model.predict()` để tạo dự đoán cho cả train và test. Sau đó:

- **Accuracy** = số dự đoán đúng / tổng số mẫu.
- **Precision của TV Show**: trong các nội dung model đoán là TV Show, bao nhiêu nội dung thật sự là TV Show.
- **Recall của TV Show**: trong toàn bộ TV Show thật, model tìm thấy được bao nhiêu.
- **F1-score**: trung bình điều hòa giữa precision và recall; chỉ cao khi cả hai cùng tốt.
- **Macro average**: tính metric riêng cho mỗi lớp rồi lấy trung bình, nên Movie và TV Show có vai trò ngang nhau.
- **Weighted average**: lấy trung bình có tính đến số lượng mẫu của từng lớp.

`accuracy_gap = train_accuracy - test_accuracy` so sánh khả năng ghi nhớ train set với khả năng tổng quát hóa. Notebook dùng ngưỡng minh họa 0.10: chênh lệch lớn hơn 10 điểm phần trăm là dấu hiệu cần kiểm tra overfitting. Đây là quy ước thực hành, không phải định luật tuyệt đối.

```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)
accuracy_gap = train_accuracy - test_accuracy

results = pd.DataFrame({
    'Tập dữ liệu': ['Train', 'Test'],
    'Accuracy': [train_accuracy, test_accuracy]
})
display(results.style.format({'Accuracy': '{:.3f}'}))

print('Classification report trên test set:\n')
print(classification_report(
    y_test, y_test_pred, target_names=['Movie', 'TV Show'], digits=3
))
print(f'Chênh lệch Train - Test: {accuracy_gap:.3f}')
if accuracy_gap > 0.10:
    print('Cảnh báo: model có dấu hiệu overfitting.')
else:
    print('Không có dấu hiệu overfitting nghiêm trọng theo ngưỡng 0.10.')
```

```python
# Confusion matrix cho biết model đúng/sai theo từng lớp.
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_test_pred,
    display_labels=['Movie', 'TV Show'],
    cmap='Reds', values_format='d', ax=ax
)
ax.set_title('Confusion Matrix - Logistic Regression')
plt.tight_layout()
plt.show()
```

### Cách đọc confusion matrix

Hàng biểu diễn nhãn thật, cột biểu diễn nhãn model dự đoán:

- Ô Movie → Movie: Movie được dự đoán đúng.
- Ô Movie → TV Show: Movie bị dự đoán nhầm thành TV Show.
- Ô TV Show → Movie: TV Show bị bỏ sót.
- Ô TV Show → TV Show: TV Show được dự đoán đúng.

Hai ô trên đường chéo là dự đoán đúng; hai ô ngoài đường chéo là lỗi.

## 6. Cross-validation

Một lần chia train/test có thể may mắn hoặc không may tùy các dòng rơi vào test set. Cross-validation kiểm tra độ ổn định bằng cách:

1. Chia toàn bộ dữ liệu thành 5 phần (fold).
2. Mỗi lượt dùng 4 phần để train và 1 phần để validation.
3. Lặp 5 lượt để mỗi phần được làm validation đúng một lần.
4. Báo F1-macro của từng lượt, trung bình và độ lệch chuẩn.

F1-macro đánh giá hai lớp ngang nhau, tránh để lớp Movie đông hơn chi phối kết quả. `n_jobs=1` chạy tuần tự để tương thích ổn định với Windows. Cross-validation chỉ dùng để đánh giá; object `model` phía trên vẫn là model đã fit trên `X_train`.

```python
cv_scores = cross_val_score(model, X, y, cv=5, scoring='f1_macro', n_jobs=1)
print('F1-macro từng fold:', np.round(cv_scores, 3))
print(f'F1-macro trung bình: {cv_scores.mean():.3f} ± {cv_scores.std():.3f}')
```

## 7. Dự đoán một nội dung mới

`new_content` phải có đúng bốn cột đầu vào đã dùng khi train. Pipeline tự áp dụng lại TF-IDF, One-Hot Encoding và StandardScaler; người dùng không cần xử lý thủ công.

- `model.predict(new_content)[0]` trả về lớp 0 hoặc 1 của dòng đầu tiên.
- `model.predict_proba(new_content)[0]` trả về `[P(Movie), P(TV Show)]`.
- `label_map` đổi số 0/1 trở lại tên lớp dễ đọc.

Xác suất này là mức tự tin của model theo dữ liệu đã học, không phải sự thật tuyệt đối. Ví dụ dùng cụm *several episodes*, nên nội dung có xu hướng được dự đoán là TV Show.

```python
new_content = pd.DataFrame([{
    'description': 'A detective investigates a mystery across several episodes.',
    'rating': 'TV-MA',
    'country': 'United States',
    'release_year': 2024
}])

predicted_class = model.predict(new_content)[0]
predicted_probability = model.predict_proba(new_content)[0]
label_map = {0: 'Movie', 1: 'TV Show'}

print('Dự đoán:', label_map[predicted_class])
print(f'Xác suất Movie: {predicted_probability[0]:.1%}')
print(f'Xác suất TV Show: {predicted_probability[1]:.1%}')
```

## 8. Kết luận

Logistic Regression là baseline phù hợp cho bài toán phân loại Movie/TV Show. Trong lần chạy kiểm chứng hiện tại:

- Train accuracy khoảng **85,4%**.
- Test accuracy khoảng **76,0%**.
- Test F1-macro khoảng **0,743**.
- F1-macro cross-validation khoảng **0,742 ± 0,023**.
- Chênh lệch train–test khoảng **0,093**, dưới ngưỡng cảnh báo minh họa 0,10.

Model nhận diện TV Show với recall cao nhưng precision thấp hơn Movie, tức là tìm được nhiều TV Show thật nhưng cũng gán nhầm một số Movie thành TV Show. Đây là model học từ metadata, không phải tiêu chuẩn chính thức để xác định loại nội dung.
