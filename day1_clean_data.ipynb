```python
import pandas as pd
df = pd.read_csv("netflix_titles.csv")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
df.info()
```

```python
print("Missing values theo cột:")
print(df.isnull().sum())

print("\nBản ghi trùng lặp:", df.duplicated().sum())

# date_added đang là text, cần parse thành ngày tháng
print("\nMẫu date_added:", df["date_added"].head(3).tolist())
```

```python
# 1) Ngày tháng: chuyển sang datetime thật
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

# 2) Cột text thiếu dữ liệu -> điền "Unknown" thay vì xoá dòng
#    (xoá dòng sẽ mất luôn thông tin title/type/rating vẫn hợp lệ)
for col in ["director", "cast", "country"]:
    df[col] = df[col].fillna("Unknown")

# 3) duration: tách số và đơn vị riêng vì Movie/TV Show khác đơn vị
df["duration_value"] = df["duration"].str.extract(r"(\d+)").astype(float)
df["duration_unit"] = df["duration"].str.extract(r"([a-zA-Z]+)")

print(df.isnull().sum())
print(df.dtypes)
```

```python
df.to_csv("netflix_titles_clean.csv", index=False)
```
